```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer, KNNImputer

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

print("Original Dataset Shape:", df.shape)
df.head()
```

```python
categorical_col = 'Sex'
print("Selected Categorical Column:", categorical_col)
print(df[categorical_col].value_counts())
```

```python
encoded = pd.get_dummies(df[categorical_col], prefix=categorical_col, dtype=int)

print("New Encoded Columns:")
print(encoded.head())
```

```python
col_index = df.columns.get_loc(categorical_col)

df_clean = df.drop(columns=[categorical_col])

for i, col in enumerate(encoded.columns):
    df_clean.insert(col_index + i, col, encoded[col])

print("Columns after encoding and insertion:")
print(df_clean.columns.tolist())
df_clean.head()
```

```python
missing = df_clean.isnull().sum()
missing_percent = (missing / len(df_clean)) * 100

missing_df = pd.DataFrame({
    'Missing Values': missing,
    'Percentage (%)': missing_percent.round(2)
})

print("Missing Values Analysis:")
print(missing_df[missing_df['Missing Values'] > 0])
```

```python
df_final = df_clean.copy()

if 'Cabin' in df_final.columns:
    df_final = df_final.drop(columns=['Cabin'])

imputer_age = SimpleImputer(strategy='median')
df_final['Age'] = imputer_age.fit_transform(df_final[['Age']]).ravel()

imputer_emb = SimpleImputer(strategy='most_frequent')
df_final['Embarked'] = imputer_emb.fit_transform(df_final[['Embarked']]).ravel()

print("✅ Missing values after imputation:", df_final.isnull().sum().sum())
print("\nFinal check - Embarked missing values:", df_final['Embarked'].isnull().sum())
```
